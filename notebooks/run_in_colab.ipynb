## 3. Read the report

```python
from pathlib import Path
from IPython.display import Image, Markdown, display

out = Path("output") / MONTH
report = (out / "report.md").read_text()
# Charts are shown separately below, because Colab can't load the report's relative image paths.
display(Markdown("\n".join(line for line in report.splitlines() if not line.startswith("!["))))
for chart in sorted((out / "charts").glob("*.png")):
    display(Image(filename=str(chart)))
```
